# Interactive BMI Calculator

```python
import ipywidgets as widgets
from IPython.display import display

# Create the input controls
name_input = widgets.Text(
    description="Name:",
    placeholder="Enter your name"
)

weight_input = widgets.FloatText(
    description="Weight (lb):",
    value=120,
    step=0.1
)

weight_slider = widgets.FloatSlider(
    description="Weight (lb):",
    value=120,
    min=0,
    max=500,
    step=0.1,
    readout_format=".1f"
)

height_input = widgets.FloatText(
    description="Height (in):",
    value=65,
    step=0.1
)

height_slider = widgets.FloatSlider(
    description="Height (in):",
    value=65,
    min=0,
    max=100,
    step=0.1,
    readout_format=".1f"
)

# Keep the numeric fields and sliders synchronized
weight_link = widgets.link((weight_input, 'value'), (weight_slider, 'value'))
height_link = widgets.link((height_input, 'value'), (height_slider, 'value'))

# Observer function to round the values of the sliders and text inputs
def round_value(change):
    new_value = change["new"]
    rounded_value = round(new_value, 1)

    # Prevent repeated updates once the value is already rounded 
    if new_value != rounded_value:
        change["owner"].value = rounded_value

# Connect the observer function to the sliders and text inputs
weight_input.observe(round_value, names='value')
height_input.observe(round_value, names='value')

# Create the calculation button and a dedicated results area
calculate_button = widgets.Button( 
    description="Calculate BMI",
    button_style="success")

output = widgets.Output()


# Calculate and display the BMI when the button is clicked
def calculate_bmi(button):
    with output:
        output.clear_output()

        name = name_input.value.strip()
        weight = weight_input.value
        height = height_input.value

        # Reject invalid measurements and prevent division by zero
        if weight <= 0 or height <= 0:
            print("Please enter a weight and height greater than zero.")
            return

        bmi = (weight * 703) / (height * height)

        # Calculate the estimated weight range corresponding to a healthy-BMI category
        lower_weight_boundary = (18.5 * (height * height)) / 703
        upper_weight_boundary = (25 * (height * height)) / 703

        # Determine BMI category
        if bmi < 18.5:
            category = "Underweight"
        elif bmi < 25: # it represents the range 18.5 <= bmi < 25
            category = "Healthy weight"
        elif bmi < 30:
            category = "Overweight"
        elif bmi < 35:
            category = "Obesity — Class 1"
        elif bmi < 40:
            category = "Obesity — Class 2"
        else:
            category = "Obesity — Class 3"

        # Describe the entered weight's position relative to that range

        if bmi < 18.5:
            range_position = "below"
        elif bmi < 25:
            range_position = "within"
        else:
            range_position = "above"

        weight_message = (
            f"For adults, a BMI between 18.5 and 24.9 is considered within "
            f"the healthy-weight range.  \nBased on your height, this corresponds "
            f"to approximately {lower_weight_boundary:.1f}–"
            f"{upper_weight_boundary:.1f} lb. "
            f"Your weight falls {range_position} this range."
        )

        greeting = f"{name}, your" if name else "Your"

        print(f"{greeting} BMI is: {bmi:.2f}")
        print(f"BMI Category: {category}")
        print(weight_message)

# Connect the button to the function
calculate_button.on_click(calculate_bmi)

# Display weight and height in horizontal rows
measurements = widgets.HBox([
    weight_input,
    height_input
])

sliders = widgets.HBox([
    weight_slider,
    height_slider
])

# Display everything in a vertical stack
display(widgets.VBox([ 
    name_input,
    measurements,
    sliders,
    calculate_button,
    output
]))
```
